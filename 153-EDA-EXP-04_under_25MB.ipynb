```python
import pandas as pd
import plotly.express as px
import json
import urllib.request
```

```python
# ---------- 4A: World Map ----------

world_df = pd.DataFrame({
    'country': [
        'India', 'United States', 'China', 'Brazil',
        'Germany', 'Australia', 'South Africa',
        'Japan', 'Canada', 'Russia'
    ],
    'population_millions': [
        1428, 335, 1412, 216, 84,
        26, 60, 124, 39, 144
    ]
})
```

```python
fig_world = px.choropleth(
    world_df,
    locations='country',
    locationmode='country names',
    color='population_millions',
    hover_name='country',
    color_continuous_scale='Viridis',
    title='World Population by Country (Sample Data)'
)

fig_world.update_geos(
    showcoastlines=True,
    showland=True
)

fig_world.show()
```

```python
# ---------- 4B: India State-wise Map ----------

india_states_geojson_url = (
    'https://raw.githubusercontent.com/geohacker/india/master/'
    'state/india_telengana.geojson'
)

with urllib.request.urlopen(
    india_states_geojson_url
) as response:
    india_states_geo = json.load(response)

state_df = pd.DataFrame({
    'state': [
        'Maharashtra', 'Uttar Pradesh',
        'Tamil Nadu', 'Karnataka',
        'Gujarat', 'West Bengal',
        'Rajasthan', 'Kerala'
    ],
    'literacy_rate': [
        82.3, 67.7, 80.1, 75.4,
        78.0, 76.3, 66.1, 94.0
    ]
})
```

```python
fig_states = px.choropleth(
    state_df,
    geojson=india_states_geo,
    featureidkey='properties.NAME_1',
    locations='state',
    color='literacy_rate',
    color_continuous_scale='Plasma',
    hover_name='state',
    title='Literacy Rate by Indian State (Sample Data)'
)

fig_states.update_geos(
    fitbounds='locations',
    visible=False
)

fig_states.show()
```

```python
# ---------- 4C: India District-level Map ----------

district_geojson_url = (
    'https://raw.githubusercontent.com/geohacker/india/master/'
    'district/india_district.geojson'
)

with urllib.request.urlopen(
    district_geojson_url
) as response:
    india_district_geo = json.load(response)

district_df = pd.DataFrame({
    'district': [
        'Pune', 'Chennai',
        'Bengaluru Urban',
        'Lucknow', 'Kolkata'
    ],
    'population_lakhs': [
        94.3, 46.5, 96.2, 45.9, 44.9
    ]
})
```

```python
fig_district = px.choropleth(
    district_df,
    geojson=india_district_geo,
    featureidkey='properties.NAME_2',
    locations='district',
    color='population_lakhs',
    color_continuous_scale='Turbo',
    hover_name='district',
    title='Population by District (Sample Data)'
)

fig_district.update_geos(
    fitbounds='locations',
    visible=False
)

fig_district.show()
```
